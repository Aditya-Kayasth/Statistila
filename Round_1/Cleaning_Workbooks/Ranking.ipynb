```python
import pandas as pd
```

```python
ranking = pd.read_csv('/content/drive/MyDrive/Statistella/ranking.csv')
ranking[:5]
```

```python
ranking[['HOME_WINS', 'HOME_LOSSES']] = ranking['HOME_RECORD'].str.split('-', expand=True)
```

```python
ranking[['ROAD_WINS', 'ROAD_LOSSES']] = ranking['ROAD_RECORD'].str.split('-', expand=True)
```

```python
ranking['STANDINGSDATE'] = pd.to_datetime(ranking['STANDINGSDATE'])
```

```python
ranking.drop(columns=['HOME_RECORD', 'ROAD_RECORD'], inplace=True)
```

```python
ranking.drop(['LEAGUE_ID'], axis = 1, inplace = True)
```

```python
ranking['HOME_WINS'] = ranking['HOME_WINS'].astype(int)
ranking['HOME_LOSSES'] = ranking['HOME_LOSSES'].astype(int)
ranking['ROAD_WINS'] = ranking['ROAD_WINS'].astype(int)
ranking['ROAD_LOSSES'] = ranking['ROAD_LOSSES'].astype(int)
```
